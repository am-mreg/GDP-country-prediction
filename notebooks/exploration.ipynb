```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/gdp_1960_2020.csv")

# Shfaq disa rreshta
print(df.head())

# Statistika bazë
print(df.describe())

# GDP mesatare sipas shtetit
avg_gdp = df.groupby("Country")["GDP"].mean().sort_values(ascending=False).head(10)
plt.figure(figsize=(10,6))
sns.barplot(x=avg_gdp.values, y=avg_gdp.index)
plt.title("Top 10 Countries by Average GDP")
plt.show()
```
